```python
# ============================================
# WORLD CUP PREDICTOR (Fallback Version)
# No external dependencies needed!
# ============================================

import streamlit as st
import matplotlib.pyplot as plt
import numpy as np

# Simple predictor based on run rate and wickets
def predict_win(target, current_score, wickets_down, overs_done):
    # Sanity checks
    if target <= 0 or current_score <= 0 or overs_done <= 0:
        return 50
    
    overs_remaining = max(20 - overs_done, 0.1)
    required_run_rate = (target - current_score) / overs_remaining
    current_run_rate = current_score / overs_done if overs_done > 0 else 0
    
    # Wicket impact
    wicket_penalty = wickets_down * 5  # Each wicket reduces probability
    base_prob = 50
    
    # Run rate impact
    if current_run_rate > required_run_rate * 1.3:
        base_prob = 85
    elif current_run_rate > required_run_rate * 1.1:
        base_prob = 70
    elif current_run_rate > required_run_rate * 0.9:
        base_prob = 50
    elif current_run_rate > required_run_rate * 0.7:
        base_prob = 30
    else:
        base_prob = 15
    
    # Apply wicket penalty
    final_prob = max(base_prob - wicket_penalty, 5)
    return final_prob

# Test the predictor
print("🔮 World Cup Predictor (Fallback Version)")
print("-" * 40)

target = 180
current_score = 120
wickets_down = 5
overs_done = 15

win_prob = predict_win(target, current_score, wickets_down, overs_done)
print(f"Target: {target}")
print(f"Current Score: {current_score}/{wickets_down}")
print(f"Overs: {overs_done}/20")
print(f"Win Probability: {win_prob:.1f}%")
```
